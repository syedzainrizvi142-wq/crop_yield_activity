```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
#Step 1 is to load the dataset
df_ecom = pd.read_csv("pakistani_ecommerce_dataset.csv")
```

```python
# Q1: Which is the most popular product category?
category_counts = df_ecom["product_category"].value_counts()
print("\nQ1 - Most popular product category:")
print(category_counts)
print("Top category:", category_counts.index[0])
```

```python
# Q2: What is the average order value?
aov = df_ecom["order_value"].mean()
print(f"\nQ2 - Average order value: {aov:.2f}")
```

```python
# Q3: Which city has the highest number of orders?
city_counts = df_ecom["city"].value_counts()
print("\nQ3 - Orders by city:")
print(city_counts)
print("Highest order city:", city_counts.index[0])
```

```python
# Q4: What is the most used payment method?
payment_counts = df_ecom["payment_method"].value_counts()
print("\nQ4 - Payment methods:")
print(payment_counts)
print("Most used payment method:", payment_counts.index[0])
```

```python
# Q5: Draw a bar chart showing average order value by city
aov_by_city = (
    df_ecom.groupby("city")["order_value"].mean().sort_values(ascending=False)
)
print("\nQ5 - Average order value by city:")
print(aov_by_city)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_ecom,
    x="city",
    y="order_value",
    estimator="mean",
    order=aov_by_city.index,
)
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.savefig("q5_aov_by_city.png", dpi=300, bbox_inches="tight")
plt.show()
```
