```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
sns.set_style("whitegrid")
sns.set_palette("pastel")
```

```python
crop_df = pd.read_csv("crop_yield.csv")
```

```python
print("\n--- First 10 Rows ---")
print(crop_df.head(10))
```

```python
print("\n--- Missing Values Before Cleaning ---")
print(crop_df.isnull().sum())
```

```python
print("\n--- Data Types Before Cleaning ---")
print(crop_df.dtypes)
```

```python
print("\n--- Numeric Summary ---")
print(crop_df.describe())
```

```python
print("\n--- Object Summary ---")
print(crop_df.describe(include=["object"]))
```

```python
crop_df = crop_df.dropna(subset=["YIELD_KG"]).copy()
```

```python
crop_df["AREA_ACRES"] = crop_df["AREA_ACRES"].fillna(
    crop_df["AREA_ACRES"].mean()
)
```

```python
crop_df["DISTRICT"] = crop_df["DISTRICT"].fillna("Unknown")
```

```python
crop_df["FERTILIZER_USED"] = crop_df["FERTILIZER_USED"].fillna("No")
```

```python
crop_df["YIELD_VALUE"] = crop_df["YIELD_VALUE"].fillna(
    crop_df["YIELD_VALUE"].mean()
)
```

```python
crop_df["HARVEST_DATE"] = pd.to_datetime(
    crop_df["HARVEST_DATE"], format="%d/%m/%Y", errors="coerce"
)
```

```python
crop_df["FARM_ID"] = (
    crop_df["FARM_ID"].astype(str).str.replace("F", "", regex=False).astype(int)
)
```

```python
print("\n--- Missing Values After Cleaning ---")
print(crop_df.isnull().sum())
```

```python
print("\n--- Data Types After Cleaning ---")
print(crop_df.dtypes)
```

```python
yield_by_crop = crop_df.groupby("CROP_TYPE").agg({"YIELD_KG": "sum"})
print("\n--- Total Yield by Crop Type (KG) ---")
print(yield_by_crop)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=yield_by_crop.index, y=yield_by_crop["YIELD_KG"])
plt.title("Total Yield by Crop Type (KG)")
plt.xlabel("Crop Type")
plt.ylabel("Yield (KG)")
plt.grid(True, alpha=0.3)
plt.savefig("crop_yield_by_crop.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
pivot_yield = crop_df.pivot_table(
    values="YIELD_KG", index="HARVEST_DATE", aggfunc="sum"
)
print("\n--- Pivot Table: Total Yield by Harvest Date ---")
print(pivot_yield.head(10))
```

```python
plt.figure(figsize=(12, 6))
plt.plot(pivot_yield.index, pivot_yield["YIELD_KG"])
plt.title("Yield Trend Over Time (KG)")
plt.xlabel("Harvest Date")
plt.ylabel("Yield (KG)")
plt.grid(True, alpha=0.3)
plt.savefig("crop_yield_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
yearly_yield = pivot_yield.groupby(pivot_yield.index.year)["YIELD_KG"].sum()

plt.figure(figsize=(12, 6))
plt.plot(
    yearly_yield.index,
    yearly_yield.values,
    marker="o",
    linewidth=2.5,
    markersize=7,
    color="#5B9BD5",
    label="Total Yield (KG)",
)
plt.gca().yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
plt.title("Yield Trend Over Time (KG)")
plt.xlabel("Harvest Year")
plt.ylabel("Yield (KG)")
plt.xticks(yearly_yield.index)
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig("crop_yield_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```
